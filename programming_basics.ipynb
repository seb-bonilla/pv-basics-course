## 1. Variables

A variable stores a value so you can reuse it later.

Think of it like a labeled box. The label is the name, and the value is what is stored.

Key idea: a variable is just a name that points to data.

```python
temperature = 25
name = "Asha"

print(name)
print(temperature)
```

## 2. Data types

Data can be different kinds: numbers, text, and true/false values.

Common types:
- `int` for whole numbers
- `float` for decimals
- `str` for text
- `bool` for True/False

Type matters because the same symbol can behave differently for different types.

```python
age = 21
price = 4.99
label = "data"
is_ready = True

print(type(age))
print(type(price))
print(type(label))
print(type(is_ready))
```

Mini-task: run the cell, then change one value so its type changes.

```python
# Change one value below, then run again.
age = 21
label = "data"  # try changing this
print(type(age))
print(type(label))
```

## 3. Expressions

An expression is any piece of code that gives a value.

Examples:
- `2 + 3`
- `x * 10`
- `"hi" + "!"`

Expressions can use variables, numbers, and operators.

```python
x = 5
y = 2

print(x + y)
print(x * y)
print(x ** y)
```

```python
# Predict first, then run.
x = 7
y = 3
print(x + y)
print(x - y)
print(x * y)
```

## 4. Functions

A function is a named block of code that takes input and gives output.

Why they help:
- they reduce repeated code
- they make programs easier to read
- they let you reuse logic

Most functions follow this pattern: input → process → output.

```python
def square(x):
    return x * x

print(square(3))
print(square(10))
```

Mini-task: write a function called `double(x)` that returns twice the input.

Test it with a few numbers.

```python
def double(x):
    return x * 2

print(double(4))
print(double(9))
```

## 5. Methods

A method is a function that belongs to an object.

Example: a string can do things like `upper()` or `replace()`.

Think: the object carries its own tools.

```python
text = "hello world"

print(text.upper())
print(text.replace("world", "data"))
```

Mini-task: start with a string and use one method to change it.

Try `lower()`, `strip()`, or `replace()`.

```python
text = "  Data Science  "
print(text.strip())
print(text.lower())
```

## 6. If statements

An `if` statement lets the program make a decision.

It checks a condition:
- if true, run this code
- if false, skip it or choose another path

This is how code reacts to different situations.

```python
score = 82

if score >= 50:
    print("Pass")
else:
    print("Try again")
```


Mini-task: change `score` so the result switches between the two branches.

## 7. Loops

A loop repeats code.

Use a loop when you want to do the same thing many times, or when you need to go through a list of items.

Common loop: `for`.

```python
for i in range(5):
    print("Step", i)
```

Mini-task: make the loop print from 1 to 10 instead of 0 to 4.

```python
for i in range(1, 11):
    print(i)
```

## 8. Collections

Collections store more than one item.

Two very common ones:
- list: ordered items
- dictionary: key-value pairs

Use lists for sequences and dictionaries for named values.

```python
numbers = [2, 4, 6]
student = {"name": "Mina", "age": 20}

print(numbers[0])
print(student["name"])
```

Mini-task: add one more item to the list and one more key-value pair to the dictionary.

```python
numbers = [2, 4, 6, 8]
student = {"name": "Mina", "age": 20, "grade": "A"}

print(numbers[2])
print(student)
```

## 9. Modules and libraries

A module is code someone else wrote that you can reuse.

This saves time and avoids rewriting common tools. In Python, you use `import` to bring a module into your notebook.

```python
import math

print(math.sqrt(16))
print(math.pi)
```

Mini-task: import `random` and print one random integer between 1 and 10.

```python
import random

print(random.randint(1, 10))
```

## 10. Debugging

Debugging means finding and fixing problems in code.

Good debugging habits:
- read the error message
- check the line number
- inspect variable values
- test small steps instead of guessing

Errors are normal. They are part of learning.

```python
x = "5"
y = 2

# This line will fail because text and numbers are handled differently.
# Try to understand the error message before changing the code.
print(x + y)
```
