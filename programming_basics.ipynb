## 2. Data types

Data can be different kinds: numbers, text, and true/false values.

Common types:
- `int` for whole numbers

## 4. Functions

A function is a named block of code that takes input and gives output.

Why they help:
- they reduce repeated code
- they make programs easier to read
- they let you reuse logic

Most functions follow this pattern: input → process → output.

```python
def square(x):
    return x * x

print(square(3))
print(square(10))
```

## 6. If statements

An `if` statement lets the program make a decision.

It checks a condition:
- if true, run this code
- if false, skip it or choose another path

This is how code reacts to different situations.

```python
score = 82

if score >= 50:
    print("Pass")
else:
    print("Try again")
```

Let's now use that random function and the pltoing library to get a whole bunch of random numbers.

```python
import random
import matplotlib.pyplot as plt

# List to store random values
random_values = []

# Generate 100 random values
for _ in range(100):
    random_values.append(random.random()) # Generates a float between 0.0 and 1.0

# Plot the 100 random values
plt.figure(figsize=(4, 3))
plt.plot(random_values, marker='o', linestyle='-', markersize=4)
plt.title('100 Random Values')
plt.xlabel('Index')
plt.ylabel('Random Value')
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(4, 3))
plt.hist(random_values, bins=10, edgecolor='black') # Using 10 bins as a default
plt.title('Histogram of 100 Random Values')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()
```
